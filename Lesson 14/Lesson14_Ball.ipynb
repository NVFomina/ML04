```python
import numpy as np
from PIL import Image
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
img = np.array(Image.open('шарик2.jpg'))
img.shape
```

```python
img[0, 0]
```

```python
def grb_to_gray_scale(image):
    r, g, b = image[:,:,0], image[:,:,1], image[:,:,2]
    gray = 0.2989 * r + 0.5870 * g + 0.1140 * b
    return gray.astype(np.uint8)

image_gray = grb_to_gray_scale(img)
Image.fromarray(image_gray.astype(np.uint8))
```

```python
gray = img @ [0.2989, 0.5870,  0.1140] 
Image.fromarray(gray.astype(np.uint8))
```

```python
sns.histplot(image_gray.ravel())
```

```python
# Бинарная маска
mask = image_gray > 100  # bool array (H, W)
```

```python
Image.fromarray(np.stack([np.where(mask, 255,img[...,0]),
np.where(mask, 255,img[...,1]),
np.where(mask, 255,img[...,2])],axis=-1).astype(np.uint8))
```
